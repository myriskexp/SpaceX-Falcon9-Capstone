```python
import pandas as pd

# URL del dataset proporcionado por IBM
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'

# Leer el CSV directamente desde la URL usando pandas
spacex_df = pd.read_csv(URL)

# Mostrar las primeras filas para confirmar la carga exitosa
spacex_df.head()
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #005a9c; border-radius: 5px; font-family: sans-serif;">
    <h2 style="margin-top: 0; color: #005a9c;">Filtrar y Agrupar Sitios de Lanzamiento</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Seleccionamos las columnas relevantes para nuestro análisis espacial: el nombre del sitio, sus coordenadas (latitud y longitud) y el estado del lanzamiento (class). A continuación, usamos <code>groupby</code> para obtener una lista única de los sitios de lanzamiento y sus coordenadas, tomando el primer registro de cada grupo.</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">El resultado será un DataFrame pequeño que mostrará únicamente los sitios de lanzamiento únicos (usualmente 4) junto con sus valores de <code>Lat</code> y <code>Long</code>. Verifica que no haya sitios duplicados en esta tabla, ya que usaremos estas coordenadas exactas para colocar los marcadores en el mapa.</p>
</div>

```python
# Select relevant sub-columns: `Launch Site`, `Lat`, `Long`, `class`
spacex_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']]

# Group by Launch Site to get unique coordinates
launch_sites_df = spacex_df.groupby(['Launch Site'], as_index=False).first()

# Keep only the columns needed for marking the map
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]

# Display the resulting dataframe
launch_sites_df
```

    <h2 style="margin-top: 0; color: #005a9c;">Agregar Marcadores al Mapa Base</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Añadimos dos elementos visuales a nuestro mapa en las coordenadas del Centro Espacial Johnson de la NASA. Primero, creamos un <code>folium.Circle</code> (un círculo de color naranja con un radio de 1000 metros) que incluye un mensaje emergente (popup). <div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #005a9c; border-radius: 5px; font-family: sans-serif;">
    <h2 style="margin-top: 0; color: #005a9c;">Tarea 1: Marcar todos los sitios de lanzamiento</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Reiniciamos el mapa con un nivel de zoom menor (5) para obtener una vista continental. Luego, implementamos un bucle <code>for</code> que itera sobre cada fila del DataFrame <code>launch_sites_df</code>. Por cada sitio de lanzamiento, extraemos sus coordenadas (Latitud y Longitud) y agregamos dinámicamente un círculo con un popup y un marcador de texto utilizando la misma técnica aplicada previamente para el centro de la NASA.</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">El mapa actualizado mostrará marcadores naranjas para todos los sitios de lanzamiento presentes en el dataset. Al observar el mapa, podrás empezar a identificar patrones geográficos comunes, como su proximidad a la costa o su ubicación relativa al ecuador.</p>

```python
# Initialize the map with a wider zoom to see all of the US
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# For each launch site, add a Circle object based on its coordinate (Lat, Long) values. 
# In addition, add Launch site name as a popup label
for index, row in launch_sites_df.iterrows():
    # Coordenadas actuales del bucle
    site_coordinate = [row['Lat'], row['Long']]
    site_name = row['Launch Site']
    
    # Crear un círculo coloreado
    circle = folium.Circle(
        site_coordinate, 
        radius=1000, 
        color='#d35400', 
        fill=True
    ).add_child(folium.Popup(site_name))
    
    # Crear un marcador con el nombre del sitio
    marker = folium.map.Marker(
        site_coordinate,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % site_name,
        )
    )
    
    # Añadir los elementos al mapa
    site_map.add_child(circle)
    site_map.add_child(marker)

# Display the map
site_map
```

    <h2 style="margin-top: 0; color: #005a9c;">Asignar Colores a los Marcadores</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Definimos una función en Python que evalúa el resultado de un lanzamiento. Si el valor es 1 (éxito), devuelve 'green'; si es 0 (fallo), devuelve 'red'. Utilizaremos el método <code>apply</code> de pandas para ejecutar esta función sobre la columna <code>class</code> y almacenar los resultados en una nueva columna llamada <code>marker_color</code> dentro de nuestro DataFrame <code>spacex_df</code>.</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">La celda mostrará las últimas 10 filas del DataFrame actualizado.
```python
# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red

def assign_marker_color(launch_outcome):
    if launch_outcome == 1:
        return 'green'
    else:
        return 'red'
    
# Create a new column in spacex_df dataframe called marker_color
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# Check the tail of the dataframe to confirm the new column
spacex_df.tail(10)
```

Por cada lanzamiento, creamos un <code>folium.Marker</code> en sus coordenadas exactas, asignándole un icono cuyo color interno (<code>icon_color</code>) lee dinámicamente la columna <code>marker_color</code> que creamos en el paso anterior.     <h2 style="margin-top: 0; color: #005a9c;">Función para Calcular Distancias</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Importamos funciones matemáticas estándar y definimos la función <code>calculate_distance</code>, la cual aplica la fórmula de Haversine. Esta fórmula nos permite calcular la distancia en línea recta (en kilómetros) entre dos puntos en una esfera, dadas sus latitudes y longitudes. Es totalmente compatible con tu entorno de Python 3.12[cite: 2].</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">La celda se ejecutará de forma silenciosa y el mensaje de confirmación aparecerá abajo. Esto significa que la función quedó guardada en la memoria, lista para ser utilizada en el siguiente paso cuando elijamos un punto de interés.</p>
</div>

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

print("Función calculate_distance definida exitosamente en memoria.")
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #005a9c; border-radius: 5px; font-family: sans-serif;">
    <h2 style="margin-top: 0; color: #005a9c;">Calcular Distancia a la Costa y Agregar Marcador</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Tomamos las coordenadas de un sitio de lanzamiento específico (por ejemplo, CCAFS SLC-40) y las coordenadas de la costa más cercana provistas en el código de ejemplo de IBM. Utilizamos la función <code>calculate_distance</code> previamente en memoria para obtener la distancia exacta en kilómetros. Finalmente, creamos un marcador en el punto costero que muestra textualmente dicha distancia utilizando un <code>DivIcon</code>.</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">La celda imprimirá la distancia calculada (aproximadamente 0.51 km) y mostrará el mapa. En la costa cercana al sitio CCAFS SLC-40, deberías ver un nuevo marcador de texto en color naranja indicando la distancia exacta. Verifica que el marcador aparece en la ubicación correcta al este del sitio de lanzamiento.</p>
</div>

```python
# Usaremos las coordenadas del sitio de lanzamiento CCAFS SLC-40 como ejemplo
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# Coordenadas de la costa más cercana sugeridas en el ejercicio
coastline_lat = 28.56367
coastline_lon = -80.57163

# Calcular la distancia usando la función definida en el paso anterior
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distancia calculada a la costa: {distance_coastline:.2f} km")

# Crear y añadir un marcador en el punto costero que muestre la distancia
distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
    )
)

site_map.add_child(distance_marker)

# Mostrar el mapa actualizado
site_map
```

    <h2 style="margin-top: 0; color: #005a9c;">Dibujar una Línea (PolyLine) hacia la Costa</h2>
    <h3 style="color: #333333;">Qué hacemos</h3>
    <p style="color: #555555;">Siguiendo las instrucciones del laboratorio de IBM[cite: 1], ahora trazaremos una línea recta entre el sitio de lanzamiento y el punto costero seleccionado. Para ello, agrupamos las coordenadas de ambos puntos en una lista y utilizamos <code>folium.PolyLine</code> para generar la línea visual. Finalmente, la agregamos a nuestro mapa base.</p>
    <h3 style="color: #333333;">Qué observar</h3>
    <p style="color: #555555;">Al visualizar el mapa interactivo, deberías ver una delgada línea azul que conecta el centro del sitio de lanzamiento CCAFS SLC-40 directamente con el marcador naranja en la costa (el cual indica la distancia calculada anteriormente). Confirma que la línea une ambos puntos con precisión.</p>
</div>

```python
# Crear una lista con las coordenadas del sitio de lanzamiento y la costa
coordinates = [
    [launch_site_lat, launch_site_lon],
    [coastline_lat, coastline_lon]
]

# Crear un objeto folium.PolyLine usando las coordenadas
lines = folium.PolyLine(
    locations=coordinates, 
    weight=2,        # Aumentamos ligeramente el grosor para mejor visibilidad
    color='blue'     # Definimos el color de la línea
)

# Añadir la línea al mapa
site_map.add_child(lines)

# Mostrar el mapa actualizado
site_map
```
