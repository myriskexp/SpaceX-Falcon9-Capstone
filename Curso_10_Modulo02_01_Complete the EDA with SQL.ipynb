<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe; border-radius: 5px;">
    <h2 style="color: #0f62fe; margin-top: 0;">Paso 1: Descripción General y Descarga del Dataset</h2>
    <p><strong>Qué hacemos:</strong></p>
    <p><strong>Contexto del conjunto de datos:</strong> SpaceX ha captado la atención mundial tras alcanzar múltiples hitos históricos. Es la única empresa privada capaz de retornar una nave espacial desde una órbita baja terrestre, logro alcanzado por primera vez en diciembre de 2010. SpaceX publicita en su portal lanzamientos del cohete Falcon 9 con un costo de 62 millones de dólares, mientras que otros proveedores cobran más de 165 millones por cada misión. Gran parte de este ahorro se debe a la capacidad de recuperar y reutilizar la primera etapa del cohete.</p>
    <p>Por lo tanto, si logramos determinar con anticipación si la primera etapa aterrizará exitosamente, podemos predecir con precisión el costo real de un lanzamiento. Esta información resulta fundamental en caso de que una compañía alterna busque competir contra SpaceX en una licitación comercial de lanzamiento espacial. Este dataset incluye los registros correspondientes a cada carga útil transportada durante las misiones espaciales de SpaceX.</p>
    <p><strong>Acción técnica:</strong> Importaremos pandas y descargaremos el archivo CSV original directamente desde el repositorio oficial de IBM Skills Network hacia un DataFrame para iniciar la fase de exploración.</p>
</div>

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe; border-radius: 5px;">
    <h2 style="color: #0f62fe; margin-top: 0;">Paso 2: Configuración de la Base de Datos SQL</h2>
    <p><strong>Qué hacemos:</strong> Prepararemos el entorno para ejecutar consultas SQL. El fragmento original asume que la extensión SQL ya está cargada y utiliza la variable <code>con</code> sin declararla previamente[cite: 1]. Para que funcione en nuestro entorno local, importaremos <code>sqlite3</code> para crear explícitamente la conexión a una base de datos llamada <code>my_data1.db</code>, cargaremos el DataFrame en la tabla <code>SPACEXTBL</code> y finalmente habilitaremos la extensión SQL de Jupyter[cite: 1].</p>
</div>

```python
import pandas as pd
import sqlite3

# 1. Crear conexión a la base de datos local SQLite
con = sqlite3.connect("my_data1.db")

# 2. Cargar los datos desde el CSV de IBM
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv"
df = pd.read_csv(url)

# 3. Guardar el DataFrame en la tabla SPACEXTBL
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False, method="multi")

# 4. Cargar la extensión SQL y conectar la magia %sql
%load_ext sql
%sql sqlite:///my_data1.db

print("Base de datos my_data1.db creada y tabla SPACEXTBL cargada con éxito.")
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe; border-radius: 5px;">
    <h2 style="color: #0f62fe; margin-top: 0;">Paso 3: Limpieza y Preparación de la Tabla SQL</h2>
    <p><strong>Qué hacemos:</strong> . En este paso, utilizaremos los comandos mágicos de SQL para crear una versión limpia de nuestros datos. Primero, nos aseguraremos de eliminar la tabla <code>SPACEXTABLE</code> si ya existe para evitar errores si ejecutas la celda varias veces. Luego, crearemos esa nueva tabla copiando la información de <code>SPACEXTBL</code>, pero filtrando cualquier registro donde la fecha (<code>Date</code>) sea nula.</p>
</div>

```python
# Eliminar la tabla SPACEXTABLE si ya existe en la base de datos
%sql DROP TABLE IF EXISTS SPACEXTABLE;

# Crear la nueva tabla filtrando las filas con fechas nulas
%sql create table SPACEXTABLE as select * from SPACEXTBL where Date is not null;
```

    <h2 style="color: #0f62fe; margin-top: 0;">Tarea 6: Boosters exitosos en Drone Ship con carga específica</h2>

    <div style="background-color: #e6f2ff; padding: 10px; border-left: 4px solid #005a9c; margin-top: 15px;">
        <p><strong>Qué observar:</strong> Al ejecutar el código, el entorno devolverá una tabla listando únicamente las versiones de los boosters que cumplen con ambos criterios.     <h2 style="color: #0f62fe; margin-top: 0;">Tarea 8: Booster Versions con la Masa Máxima de Carga Útil</h2>
    <p><strong>Qué hacemos:</strong> Estructuraremos una consulta SQL que utiliza una subconsulta (subquery). La subconsulta interna calculará el valor máximo de la columna <code>PAYLOAD_MASS__KG_</code> utilizando la función de agregación <code>MAX()</code>. La consulta externa filtrará y listará los registros de <code>Booster_Version</code> cuya carga útil sea exactamente igual a ese valor máximo encontrado.</p>
</div>

```python
%sql SELECT "Booster_Version" FROM SPACEXTABLE WHERE "PAYLOAD_MASS__KG_" = (SELECT MAX("PAYLOAD_MASS__KG_") FROM SPACEXTABLE);
```

    <h2 style="color: #0f62fe; margin-top: 0;">Tarea 9: Fallos en Drone Ship durante 2015</h2>
    <p><strong>Qué hacemos:</strong> Utilizaremos la función <code>substr()</code> para extraer el mes de la columna <code>Date</code>, tal como indica el laboratorio, debido a que SQLite no cuenta con una función nativa para obtener los nombres de los meses directamente[cite: 1]. Filtraremos los registros especificando que el año extraído sea 2015 y que el resultado del aterrizaje en la columna <code>Landing_Outcome</code> corresponda a 'Failure (drone ship)'.</p>
    <div style="background-color: #e6f2ff; padding: 10px; border-left: 4px solid #005a9c; margin-top: 15px;">
        <p><strong>Qué observar:</strong> El entorno debe retornar una tabla indicando el mes (en formato numérico 01-12), el resultado del aterrizaje, la versión del booster y el sitio de lanzamiento. Ejecuta la celda y comparte la salida obtenida para interpretarla, o escribe "siguiente" para ir a la Tarea 10[cite: 1].</p>
    </div>
</div>

```python
%sql SELECT substr("Date", 6, 2) AS Month, "Landing_Outcome", "Booster_Version", "Launch_Site" FROM SPACEXTABLE WHERE "Landing_Outcome" = 'Failure (drone ship)' AND substr("Date", 1, 4) = '2015';
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe; border-radius: 5px;">
    <h2 style="color: #0f62fe; margin-top: 0;">Tarea 10: Clasificación de Resultados de Aterrizaje</h2>
    <p><strong>Qué hacemos:</strong> Conforme al objetivo pedagógico del laboratorio[cite: 1], agruparemos los registros por la columna <code>Landing_Outcome</code> y contaremos las ocurrencias utilizando <code>COUNT()</code>. Para limitar los resultados al rango de fechas solicitado, aplicaremos la cláusula <code>WHERE</code> con el operador <code>BETWEEN</code>. Finalmente, ordenaremos los resultados de mayor a menor utilizando <code>ORDER BY</code> y la palabra clave <code>DESC</code> para completar el ranking.</p>
    <div style="background-color: #e6f2ff; padding: 10px; border-left: 4px solid #005a9c; margin-top: 15px;">
        <p><strong>Qué observar:</strong> Al ejecutar la celda, el entorno devolverá una tabla con los distintos resultados de aterrizaje y la cantidad de veces que ocurrieron entre el 2010-06-04 y el 2017-03-20, ordenados de forma descendente. Ejecuta el código y comparte la salida para finalizar esta sección del laboratorio[cite: 1].</p>
    </div>
</div>

```python
%sql SELECT "Landing_Outcome", COUNT(*) AS Outcome_Count FROM SPACEXTABLE WHERE "Date" BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY "Landing_Outcome" ORDER BY Outcome_Count DESC;
```
