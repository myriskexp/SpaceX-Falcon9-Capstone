<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">Contexto Teórico: Tipos de Órbitas Comunes</h2>
    <p><strong>Qué hacemos:</strong> Traducimos y revisamos las definiciones de las distintas órbitas objetivo para los lanzamientos de SpaceX. Comprender estas categorías es crucial para interpretar los datos de la columna <code>Orbit</code> en los próximos análisis exploratorios o modelos predictivos.</p>
    <ul>
        <li><strong>LEO (Órbita Terrestre Baja):</strong> Órbita centrada en la Tierra con una altitud de 2.000 km o menos. La mayoría de los objetos hechos por el hombre están en LEO.</li>
        <li><strong>VLEO (Órbita Terrestre Muy Baja):</strong> Órbitas con una altitud media inferior a 450 km. Proporcionan beneficios para las naves de observación de la Tierra al estar más cerca.</li>
        <li><strong>GTO (Órbita de Transferencia Geoestacionaria):</strong> Órbita elíptica utilizada para transferir satélites desde LEO hasta la órbita geoestacionaria (GEO). El perigeo es muy bajo y el apogeo alcanza unos 35.786 km.</li>
        <li><strong>SSO o SO (Órbita Heliosíncrona):</strong> Órbita casi polar donde el satélite pasa sobre cualquier punto de la superficie del planeta a la misma hora solar local media.</li>
        <li><strong>ES-L1:</strong> En los puntos de Lagrange, las fuerzas gravitacionales de dos grandes cuerpos se cancelan. L1 es uno de esos puntos entre el Sol y la Tierra.</li>
        <li><strong>HEO (Órbita Altamente Elíptica):</strong> Órbita elíptica con alta excentricidad, generalmente alrededor de la Tierra. (Nota: en la literatura de IBM a veces también se usa para referirse a órbitas por encima de la altitud geosíncrona).</li>
        <li><strong>ISS:</strong> Estación Espacial Internacional. Un satélite artificial habitable modular en órbita terrestre baja (LEO).</li>
        <li><strong>MEO (Órbita Terrestre Media):</strong> Órbitas geocéntricas con altitudes entre 2.000 km y justo por debajo de la órbita geosíncrona. Tienen comúnmente un período orbital de 12 horas.</li>
        <li><strong>GEO (Órbita Geoestacionaria):</strong> Órbita circular geosíncrona a 35.786 km sobre el ecuador de la Tierra y que sigue la dirección de rotación de esta.</li>
        <li><strong>PO (Órbita Polar):</strong> Un tipo de órbita en la que el satélite pasa por encima o casi por encima de ambos polos del cuerpo orbitado.</li>
    </ul>

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">TAREA 2: Calcular el número y ocurrencia de cada órbita</h2>
    <p><strong>Qué hacemos:</strong> Determinamos la frecuencia de cada tipo de órbita en la columna <code>Orbit</code> mediante el método <code>value_counts()</code>. Aplicamos un filtro previo al DataFrame para excluir los registros correspondientes a <code>GTO</code>, cumpliendo estrictamente con la indicación del laboratorio de omitir esta órbita de transferencia.</p>

```python
# Filtrar la órbita GTO y calcular las ocurrencias del resto de órbitas
conteo_orbitas = df[df['Orbit'] != 'GTO']['Orbit'].value_counts()

print("--- Número de lanzamientos por órbita (excluyendo GTO) ---")
print(conteo_orbitas)
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #24a148;">
    <h2 style="color: #24a148; margin-top: 0;">Interpretación de los Resultados</h2>
    <p><strong>Qué significa esto:</strong> El conteo confirma que, excluyendo la órbita de transferencia GTO, el destino más frecuente de SpaceX es la Estación Espacial Internacional (<strong>ISS</strong>) con 21 misiones, seguida por la órbita terrestre muy baja (<strong>VLEO</strong>) con 14. Esto destaca la importancia de las misiones de reabastecimiento tripuladas/no tripuladas y el despliegue de constelaciones de satélites cercanos a la Tierra dentro del conjunto de datos.</p>

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">TAREA 3: Calcular el número y ocurrencia de los resultados de las misiones</h2>
    <p><strong>Qué hacemos:</strong> Aplicamos el método <code>value_counts()</code> sobre la columna <code>Outcome</code> para contabilizar los diferentes resultados de los aterrizajes y asignamos esta serie a la variable <code>landing_outcomes</code>.</p>

```python
# Calcular los resultados de los aterrizajes usando value_counts() en la columna Outcome
landing_outcomes = df['Outcome'].value_counts()

print("--- Resultados de aterrizaje (Outcome) ---")
print(landing_outcomes)
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">Contexto Teórico: Resultados de Aterrizaje</h2>
    <p><strong>Qué hacemos:</strong> Revisamos la nomenclatura de los resultados de aterrizaje (columna <code>Outcome</code>) para comprender el éxito o fracaso de las misiones de SpaceX. Esta clasificación será clave para crear nuestra variable objetivo (target) más adelante.</p>
    <ul>
        <li><strong>True Ocean / False Ocean:</strong> Aterrizaje exitoso (True) o fallido (False) en una región específica del océano.</li>
        <li><strong>True RTLS / False RTLS:</strong> Aterrizaje exitoso (True) o fallido (False) en una plataforma terrestre (Return to Launch Site).</li>
        <li><strong>True ASDS / False ASDS:</strong> Aterrizaje exitoso (True) o fallido (False) en un barco dron (Autonomous Spaceport Drone Ship).</li>
        <li><strong>None ASDS / None None:</strong> Representan un fallo al aterrizar (generalmente porque no hubo intento o la recuperación no fue planificada).</li>
   

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">Clasificar Resultados de Aterrizaje (Bad Outcomes)</h2>
    <p><strong>Qué hacemos:</strong> Iteramos sobre los diferentes resultados de aterrizaje de la variable <code>landing_outcomes</code> para identificar la posición de cada uno. Luego, extraemos un conjunto (<code>set</code>) con los resultados que representan un fracaso utilizando los índices proporcionados por el laboratorio de IBM. Para garantizar una ejecución segura en pandas 3.0.3, utilizamos el atributo <code>.index</code> en lugar de <code>.keys()</code>.</p>

```python
# Mostrar cada resultado con su índice correspondiente
for i, outcome in enumerate(landing_outcomes.index):
    print(i, outcome)

# Agrupar los resultados fallidos (bad_outcomes) basados en sus índices
bad_outcomes = set(landing_outcomes.index[[1, 3, 5, 6, 7]])

print("\n--- Resultados categorizados como fallidos (bad_outcomes) ---")
print(bad_outcomes)
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">TAREA 4: Crear una etiqueta de resultado de aterrizaje</h2>
    <p><strong>Qué hacemos:</strong> Convertimos los resultados de aterrizaje en una variable de clasificación binaria. Iteramos sobre la columna <code>Outcome</code> para crear la lista <code>landing_class</code>: si el valor pertenece al conjunto <code>bad_outcomes</code>, le asignamos 0 (fracaso); de lo contrario, le asignamos 1 (éxito). Este paso es fundamental para definir la variable objetivo (target) que usarán los algoritmos predictivos.</p> 

```python
# Crear la variable de clasificación binaria
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]

# Visualizar los primeros 10 elementos de la lista
print("--- Primeros 10 valores de landing_class ---")
print(landing_class[:10])
```

<div style="background-color: #f4f6f7; padding: 15px; border-left: 5px solid #0f62fe;">
    <h2 style="color: #0f62fe; margin-top: 0;">Asignar la Variable Objetivo y Exportar los Datos</h2>
    <p><strong>Qué hacemos:</strong> Integramos la lista <code>landing_class</code> que creamos en el paso anterior como una nueva columna llamada <code>Class</code> en nuestro DataFrame. Luego, calculamos la tasa media de éxito de los aterrizajes (utilizando el método <code>mean()</code>) y, finalmente, exportamos el conjunto de datos limpio y preparado a un archivo CSV (<code>dataset_part_2.csv</code>) que será la base para el siguiente laboratorio.</p>

```python
# Asignar la lista de resultados a la nueva columna 'Class'
df['Class'] = landing_class

# Verificar las primeras 8 filas de la columna Class
print("--- Primeras 8 filas de la columna Class ---")
print(df[['Class']].head(8))

# Verificar las primeras 5 filas del DataFrame completo
print("\n--- Primeras 5 filas del DataFrame actualizado ---")
print(df.head(5))

# Calcular la tasa de éxito global
tasa_exito = df["Class"].mean()
print("\n--- Tasa de éxito global (Media de Class) ---")
print(tasa_exito)

# Exportar el dataset para el siguiente laboratorio
df.to_csv("dataset_part_2.csv", index=False)
print("\nArchivo 'dataset_part_2.csv' exportado exitosamente.")
```

    <p><strong>Qué significa esto:</strong> La integración de la variable objetivo fue exitosa. Al calcular la media de la columna <code>Class</code>, confirmamos que la tasa histórica de éxito de los aterrizajes en este conjunto de datos es aproximadamente del 66.67%.
